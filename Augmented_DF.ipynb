```
import pandas as pd
from statsmodels.tsa.stattools import adfuller
```

```
df = pd.read_csv("aneta.csv", index_col="DATE", parse_dates=["DATE"])
```

```
df
```

```
unrate = df["UNRATE"]
```

```
unrate.isna().sum()
```

```
result = adfuller(unrate, autolag="AIC")
print("ADF-Statistik:", result[0])
print("p-Wert:", result[1])
print("Verwendete Lags:", result[2])
print("Anzahl Beobachtungen:", result[3])
print("Kritische Werte:")
for k, v in result[4].items():
    print(f"{k}: {v:.3f}")
if result[1] < 0.05:
    print("\n-> Nullhypothese abgelehnt: Unemployment Rate ist stationär")
else:
    print("\n-> Nullhypothese nicht abgelehnt: Unemployment Rate ist nicht stationär")
```

```
result = adfuller(unrate.diff().dropna(), autolag="AIC")
print("ADF-Statistik:", result[0])
print("p-Wert:", result[1])
print("Verwendete Lags:", result[2])
print("Anzahl Beobachtungen:", result[3])
print("Kritische Werte:")
for k, v in result[4].items():
    print(f"{k}: {v:.3f}")
if result[1] < 0.05:
    print("\n-> Nullhypothese abgelehnt: Unemployment Rate ist stationär")
else:
    print("\n-> Nullhypothese nicht abgelehnt: Unemployment Rate ist nicht stationär")
```
