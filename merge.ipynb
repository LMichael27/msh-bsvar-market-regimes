```
import pandas as pd
import numpy as np
```

```
df = pd.read_csv("data/mid.csv")
```

```
df.head()
```

```
data = pd.read_excel("data/Bloomberg_US_GDP.xlsx", sheet_name="quarter")
```

```
data["Dates"] = pd.to_datetime(data["Dates"], errors="coerce")
data = data.set_index("Dates").sort_index()

data_monthly = data.resample("ME").ffill()
```

```
data_monthly.head()
```

```
df["DATE"] = pd.to_datetime(df["DATE"])
data_monthly.index = pd.to_datetime(data_monthly.index)

df["year_month"] = df["DATE"].dt.to_period("M").dt.to_timestamp("M")
data_monthly["year_month"] = data_monthly.index.to_period("M").to_timestamp("M")

merged = pd.merge(df, data_monthly[["GDP", "year_month"]], on="year_month", how="left")
merged = merged.set_index("DATE")
```

```
merged["GDP"].isna().sum()
```

```
merged = merged.dropna()
```

```
merged.to_csv("data/alles.csv")
```
