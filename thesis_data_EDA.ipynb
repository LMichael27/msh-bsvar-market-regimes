```
import pandas as pd
import numpy as np

# Daten laden
df = pd.read_csv("data/updated.csv", index_col="DATE", parse_dates=["DATE"]).sort_index()

# Auswahl der verwendeten Assets
assets = ["sp500", "LUATTRUU Index", "DXY Index", "XAU Curncy"]
df = df[assets].astype(float).dropna()

# Log-Returns
returns = np.log(df / df.shift(1)).dropna()

# Descriptive statistics
desc_stats = pd.DataFrame({
    "Mean": returns.mean(),
    "Std": returns.std(),
    "Skewness": returns.skew(),
    "Kurtosis": returns.kurtosis()
})

# Optional: annualisierte Werte
desc_stats["Mean (ann.)"] = desc_stats["Mean"] * 12
desc_stats["Std (ann.)"] = desc_stats["Std"] * np.sqrt(12)

print(desc_stats)

# Speichern für Thesis
desc_stats.to_excel("Figures/descriptive_statistics.xlsx")
```

```
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

for col in df.columns:
    plt.plot(df.index, df[col], label=col)

plt.title("Asset Price Indices (Levels)")
plt.xlabel("Time")
plt.ylabel("Index Level")
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.savefig("Figures/data/price_levels.png", dpi=150)
plt.show()
```

```
plt.figure(figsize=(12,6))

for col in returns.columns:
    plt.plot(returns.index, returns[col], label=col, alpha=0.7)

plt.title("Monthly Log Returns")
plt.xlabel("Time")
plt.ylabel("Returns")
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.savefig("Figures/data/returns.png", dpi=150)
plt.show()
```

```
rolling_vol = returns.rolling(window=12).std()

plt.figure(figsize=(12,6))

for col in rolling_vol.columns:
    plt.plot(rolling_vol.index, rolling_vol[col], label=col)

plt.title("Rolling Volatility (12 Months)")
plt.xlabel("Time")
plt.ylabel("Volatility")
plt.legend()
plt.grid(True)
plt.tight_layout()

plt.savefig("Figures/data/rolling_volatility.png", dpi=150)
plt.show()
```

```
import seaborn as sns

corr = returns.corr()

plt.figure(figsize=(6,5))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Matrix (Returns)")
plt.tight_layout()

plt.savefig("Figures/data/correlation_heatmap.png", dpi=150)
plt.show()
```
